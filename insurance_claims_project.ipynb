```python
import pandas as pd
import numpy as np

np.random.seed(42)

num_records = 1000

provinces = ["Eastern Cape", "Free State", "Gauteng", "KwaZulu-Natal", "Limpopo", "Mpumalanga", "Northern Cape", "North West", "Western Cape"]

claim_types = ["Business Insurance", "Health Insurance", "Home Insurance", "Life Insurance", "Motor Vehicle"]

claim_status = ["Approved", "Pending", "Rejected"]

# Full range: Jan to September 2025
full_dates = pd.date_range(start="2025-01-01", end="2026-06-30")

df = pd.DataFrame({
    "claim_id": range(1, num_records + 1),
    "claim_date": np.random.choice(full_dates, num_records),
    "province": np.random.choice(provinces, num_records),
    "claim_types": np.random.choice(claim_types, num_records),
    "claim_amount": np.random.randint(1000, 50000, num_records),
    "claim_status": np.random.choice(claim_status, num_records)})

df = df.sort_values("claim_date").reset_index(drop=True)

df.head()
```

```python
print("=== DATA QUALITY REPORT ===")
print("Rows, Columns:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Rows:")
print(df.duplicated().sum())
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("Minimum claim amount:", df["claim_amount"].min())
print("Maximum claim amount:", df["claim_amount"].max())
print("Average claim amount:", round(df["claim_amount"].mean(),2))
print("Provinces:")
print(df["province"].unique())
print("\nClaim Types:")
print(df["claim_types"].unique())
print("\nClaim Statuses:")
print(df["claim_status"].unique())

invalid_amounts = df[df["claim_amount"] <= 0]
print("Invalid claim amounts:", len(invalid_amounts))
```

```python
print(df["claim_status"].value_counts())
print(df["province"].value_counts())
```

```python
cleaned_df = df.copy()

print("Clean dataset created successfully.")
print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
```

```python
cleaned_df.to_csv("insurance_claims_cleaned.csv", index=False)
print("Cleaned dataset saved successfully.")
```
